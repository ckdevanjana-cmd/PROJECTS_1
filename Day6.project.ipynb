```python
# Smart Library Circulation System

books = []

while True:
    print("\n===== SMART LIBRARY =====")
    print("1. Add Book")
    print("2. View Books")
    print("3. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        book_id = input("Enter Book ID: ")
        title = input("Enter Book Title: ")
        author = input("Enter Author Name: ")
        total = int(input("Enter Total Copies: "))

        book = {
            "Book ID": book_id,
            "Title": title,
            "Author": author,
            "Total": total,
            "Available": total
        }

        books.append(book)
        print("Book added successfully!")

    elif choice == 2:
        if len(books) == 0:
            print("No books available.")
        else:
            print("\nBook List")
            for book in books:
                print("-----------------------")
                print("Book ID :", book["Book ID"])
                print("Title :", book["Title"])
                print("Author :", book["Author"])
                print("Total Copies :", book["Total"])
                print("Available Copies :", book["Available"])

    elif choice == 3:
        print("Thank You!")
        break

    else:
        print("Invalid Choice")
```
